```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
for groupID in [0, 1, 2, 3, 4]:

    df = pd.read_parquet(f"data/groups/group_{groupID}.parquet")
    keys = np.load(f"data/cluster/distances/matrix_group_{groupID}_keys.npy", allow_pickle=True)
    labels = np.load(f"data/cluster/distances/kmedoids_labels_group_{groupID}.npy")
    medoids = np.load(f"data/cluster/distances/kmedoids_medoids_group_{groupID}.npy")

    n_clusters = labels.max() + 1

    for cluster in range(n_clusters):
        plt.figure()

        # ---- cluster mask in key space ----
        cluster_mask = labels == cluster
        cluster_keys = keys[cluster_mask]

        # ---- convert keys -> DataFrame ----
        # keys are (Date, Texp)
        cluster_df = pd.DataFrame(cluster_keys, columns=["Date", "Texp"])

        # merge with original df to get w, z
        merged = cluster_df.merge(df, on=["Date", "Texp"], how="inner")

        # ---- plot cluster points ----
        plt.scatter(
            merged["z"],
            merged["w"],
            # s=20,
            alpha=0.6,
            label=f"Cluster {cluster}"
        )

        # ---- medoid ----
        medoid_idx = medoids[cluster]
        medoid_key = keys[medoid_idx]

        medoid_row = df[
            (df["Date"] == medoid_key[0]) &
            (df["Texp"] == medoid_key[1])
        ]

        if not medoid_row.empty:
            plt.scatter(
                medoid_row["z"],
                medoid_row["w"],
                # s=120,
                c="red",
                # marker="X",
                label="Medoid"
            )

        plt.xlabel("z")
        plt.ylabel("w")
        plt.title(f"Cluster {cluster}")
        plt.legend()
        plt.savefig(f"images/clusters/medoids_1/group_{groupID}_cluster_{cluster}.png")
        # plt.show()
```

```python
from scipy.interpolate import interp1d

for groupID in [0, 1, 2, 3, 4]:
    df = pd.read_parquet(f"data/groups/group_{groupID}.parquet")
    keys = np.load(f"data/cluster/distances/matrix_group_{groupID}_keys.npy", allow_pickle=True)
    labels = np.load(f"data/cluster/distances/hierachy_labels_group_{groupID}.npy")
    keys = np.array(keys)

    # ---- attach cluster labels per curve ----
    curve_df = pd.DataFrame(keys, columns=["Date", "Texp"])
    curve_df["CL"] = labels

    df = df.merge(curve_df, on=["Date", "Texp"], how="left")

    n_clusters = len(np.unique(labels))

    # global z-grid
    z_min, z_max = df["z"].min(), df["z"].max()
    common_z = np.linspace(z_min, z_max, 50)

    for cluster in np.unique(labels):
        plt.figure()

        cluster_df = df[df["CL"] == cluster]
        interpolated_curves = []

        # ---- interpolate each curve ----
        for (date, texp), sub in cluster_df.groupby(["Date", "Texp"]):

            sub = sub.dropna(subset=["z", "w"])
            sub = sub.sort_values("z")

            if len(sub) < 20:
                continue

            f = interp1d(
                sub["z"],
                sub["w"],
                kind="linear",
                bounds_error=False,
                fill_value=np.nan
            )

            w_interp = f(common_z)
            interpolated_curves.append(w_interp)

            plt.plot(sub["z"], sub["w"], color="gray", alpha=0.2)

        interpolated_curves = np.array(interpolated_curves)

        print("cluster", cluster, "curves:", interpolated_curves.shape)

        mean_w = np.nanmean(interpolated_curves, axis=0)

        plt.plot(
            common_z,
            mean_w,
            color="red",
            linewidth=3,
            label="Mean curve"
        )

        plt.xlabel("z")
        plt.ylabel("w")
        plt.title(f"Cluster {cluster}")
        plt.legend()
        # plt.show()
        plt.savefig(f"images/clusters/hierarchy/group_{groupID}_cluster_{cluster}.png")
```
